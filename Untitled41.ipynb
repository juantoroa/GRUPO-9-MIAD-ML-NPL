```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt
import seaborn as sns

# Cargar datos
# df = pd.read_excel('PROYECTO (1).xlsx', sheet_name='Hoja1 (2)')

# Preprocesamiento
# Extraer características espectrales (columnas 2-429)
X = df.iloc[:, 1:429]  # Espectros
y_ron = df.iloc[:, 429]  # RON
y_mon = df.iloc[:, 430]  # MON

# Dividir datos
X_train, X_test, y_ron_train, y_ron_test, y_mon_train, y_mon_test = train_test_split(
    X, y_ron, y_mon, test_size=0.2, random_state=42
)

# Escalar características
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# 🔹 MODELO SUPERVISADO: Random Forest para RON y MON
print("=== MODELO SUPERVISADO ===")

# Modelo para RON
rf_ron = RandomForestRegressor(n_estimators=100, random_state=42)
rf_ron.fit(X_train_scaled, y_ron_train)

# Modelo para MON
rf_mon = RandomForestRegressor(n_estimators=100, random_state=42)
rf_mon.fit(X_train_scaled, y_mon_train)

# Predicciones
y_ron_pred = rf_ron.predict(X_test_scaled)
y_mon_pred = rf_mon.predict(X_test_scaled)

# Evaluación
print(f"RON - RMSE: {np.sqrt(mean_squared_error(y_ron_test, y_ron_pred)):.3f}")
print(f"RON - R²: {r2_score(y_ron_test, y_ron_pred):.3f}")
print(f"MON - RMSE: {np.sqrt(mean_squared_error(y_mon_test, y_mon_pred)):.3f}")
print(f"MON - R²: {r2_score(y_mon_test, y_mon_pred):.3f}")

# 🔹 MODELO NO SUPERVISADO: PCA + Clustering
print("\n=== MODELO NO SUPERVISADO ===")

# Aplicar PCA
pca = PCA(n_components=0.95)  # Retener 95% de varianza
X_pca = pca.fit_transform(X_train_scaled)

print(f"Componentes PCA: {pca.n_components_}")
print(f"Varianza explicada: {np.sum(pca.explained_variance_ratio_):.3f}")

# Clustering con K-means
kmeans = KMeans(n_clusters=3, random_state=42)
clusters = kmeans.fit_predict(X_pca)

# Analizar relación clusters vs RON/MON
df_clusters = pd.DataFrame({
    'Cluster': clusters,
    'RON': y_ron_train.values,
    'MON': y_mon_train.values
})

print("\nRelación Clusters - RON/MON:")
cluster_stats = df_clusters.groupby('Cluster').agg({
    'RON': ['mean', 'std'],
    'MON': ['mean', 'std']
})
print(cluster_stats)

# 📊 Visualizaciones
plt.figure(figsize=(15, 10))

# 1. Predicciones vs Valores Reales
plt.subplot(2, 3, 1)
plt.scatter(y_ron_test, y_ron_pred, alpha=0.6)
plt.plot([y_ron_test.min(), y_ron_test.max()], [y_ron_test.min(), y_ron_test.max()], 'r--')
plt.xlabel('RON Real')
plt.ylabel('RON Predicho')
plt.title('Random Forest - RON')

plt.subplot(2, 3, 2)
plt.scatter(y_mon_test, y_mon_pred, alpha=0.6)
plt.plot([y_mon_test.min(), y_mon_test.max()], [y_mon_test.min(), y_mon_test.max()], 'r--')
plt.xlabel('MON Real')
plt.ylabel('MON Predicho')
plt.title('Random Forest - MON')

# 2. PCA - Componentes principales
plt.subplot(2, 3, 3)
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=clusters, cmap='viridis', alpha=0.6)
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.title('PCA - Clustering de Espectros')
plt.colorbar(label='Cluster')

# 3. Importancia de características (primeras 20)
plt.subplot(2, 3, 4)
feature_importance = pd.DataFrame({
    'feature': X.columns,
    'importance': rf_ron.feature_importances_
}).sort_values('importance', ascending=False).head(20)

plt.barh(feature_importance['feature'], feature_importance['importance'])
plt.xlabel('Importancia')
plt.title('Top 20 Características - RON')

# 4. Distribución de RON por cluster
plt.subplot(2, 3, 5)
sns.boxplot(data=df_clusters, x='Cluster', y='RON')
plt.title('Distribución RON por Cluster')

# 5. Distribución de MON por cluster
plt.subplot(2, 3, 6)
sns.boxplot(data=df_clusters, x='Cluster', y='MON')
plt.title('Distribución MON por Cluster')

plt.tight_layout()
plt.show()

# 🔧 Función para predecir nuevos espectros
def predict_fuel_quality(new_spectra):
    """
    Predice RON y MON para nuevos espectros
    """
    new_spectra_scaled = scaler.transform(new_spectra)

    ron_pred = rf_ron.predict(new_spectra_scaled)
    mon_pred = rf_mon.predict(new_spectra_scaled)

    # Clustering
    new_pca = pca.transform(new_spectra_scaled)
    cluster_pred = kmeans.predict(new_pca)

    return ron_pred, mon_pred, cluster_pred

# Ejemplo de uso
# new_data = [...]  # Nuevos espectros
# ron, mon, cluster = predict_fuel_quality(new_data)
```

```python
# =============================================================================
# CONFIGURACIÓN INICIAL Y CARGA DE DATOS
# =============================================================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files
from sklearn.model_selection import train_test_split, cross_val_score, GridSearchCV
from sklearn.preprocessing import StandardScaler, PolynomialFeatures
from sklearn.decomposition import PCA
from sklearn.ensemble import RandomForestRegressor, GradientBoostingRegressor, VotingRegressor
from sklearn.linear_model import ElasticNet
from sklearn.cluster import KMeans, DBSCAN
from sklearn.metrics import mean_squared_error, r2_score, silhouette_score
from sklearn.feature_selection import SelectKBest, f_regression
from sklearn.pipeline import Pipeline
import warnings
warnings.filterwarnings('ignore')

# Subir archivo
print("=== SUBIENDO ARCHIVO ===")
uploaded = files.upload()

# Cargar datos
filename = list(uploaded.keys())[0]
df = pd.read_excel(filename, sheet_name='Hoja1 (2)')
print(f"Dataset cargado: {df.shape}")

# =============================================================================
# PREPROCESAMIENTO MEJORADO
# =============================================================================
print("\n=== PREPROCESAMIENTO ===")

# Verificar y limpiar datos
print(f"Valores nulos: {df.isnull().sum().sum()}")
df = df.dropna()

# Separar características y target
X = df.drop(['RON', 'MON'], axis=1)
y_ron = df['RON']
y_mon = df['MON']

print(f"Características: {X.shape[1]}, Muestras: {X.shape[0]}")

# División train-test
X_train, X_test, y_ron_train, y_ron_test = train_test_split(X, y_ron, test_size=0.2, random_state=42)
_, _, y_mon_train, y_mon_test = train_test_split(X, y_mon, test_size=0.2, random_state=42)

# =============================================================================
# MODELO SUPERVISADO MEJORADO
# =============================================================================
print("\n=== ENTRENANDO MODELOS SUPERVISADOS MEJORADOS ===")

# Definir modelos base
models = {
    'RandomForest_RON': RandomForestRegressor(random_state=42),
    'GradientBoosting_RON': GradientBoostingRegressor(random_state=42),
    'ElasticNet_RON': ElasticNet(random_state=42),
    'RandomForest_MON': RandomForestRegressor(random_state=42),
    'GradientBoosting_MON': GradientBoostingRegressor(random_state=42),
    'ElasticNet_MON': ElasticNet(random_state=42)
}

# Pipelines mejorados
pipelines_ron = {}
pipelines_mon = {}

for name in ['RandomForest', 'GradientBoosting', 'ElasticNet']:
    pipelines_ron[name] = Pipeline([
        ('scaler', StandardScaler()),
        ('feature_selector', SelectKBest(score_func=f_regression, k=min(15, X.shape[1]))),
        ('regressor', models[f'{name}_RON'])
    ])

    pipelines_mon[name] = Pipeline([
        ('scaler', StandardScaler()),
        ('feature_selector', SelectKBest(score_func=f_regression, k=min(15, X.shape[1]))),
        ('regressor', models[f'{name}_MON'])
    ])

# Entrenar y evaluar modelos individuales
best_ron_model = None
best_mon_model = None
best_ron_score = -np.inf
best_mon_score = -np.inf

print("\n--- Evaluación de Modelos Individuales ---")
for name in ['RandomForest', 'GradientBoosting', 'ElasticNet']:
    # Para RON
    pipelines_ron[name].fit(X_train, y_ron_train)
    ron_pred = pipelines_ron[name].predict(X_test)
    ron_r2 = r2_score(y_ron_test, ron_pred)
    ron_rmse = np.sqrt(mean_squared_error(y_ron_test, ron_pred))

    # Para MON
    pipelines_mon[name].fit(X_train, y_mon_train)
    mon_pred = pipelines_mon[name].predict(X_test)
    mon_r2 = r2_score(y_mon_test, mon_pred)
    mon_rmse = np.sqrt(mean_squared_error(y_mon_test, mon_pred))

    print(f"{name}:")
    print(f"  RON - R²: {ron_r2:.4f}, RMSE: {ron_rmse:.4f}")
    print(f"  MON - R²: {mon_r2:.4f}, RMSE: {mon_rmse:.4f}")

    # Guardar mejores modelos
    if ron_r2 > best_ron_score:
        best_ron_score = ron_r2
        best_ron_model = name
    if mon_r2 > best_mon_score:
        best_mon_score = mon_r2
        best_mon_model = name

# Ensemble Voting
print("\n--- Entrenando Ensemble Model ---")
ensemble_ron = Pipeline([
    ('scaler', StandardScaler()),
    ('feature_selector', SelectKBest(score_func=f_regression, k=min(15, X.shape[1]))),
    ('regressor', VotingRegressor([
        ('rf', RandomForestRegressor(n_estimators=100, random_state=42)),
        ('gb', GradientBoostingRegressor(n_estimators=100, random_state=42)),
        ('en', ElasticNet(random_state=42))
    ]))
])

ensemble_mon = Pipeline([
    ('scaler', StandardScaler()),
    ('feature_selector', SelectKBest(score_func=f_regression, k=min(15, X.shape[1]))),
    ('regressor', VotingRegressor([
        ('rf', RandomForestRegressor(n_estimators=100, random_state=42)),
        ('gb', GradientBoostingRegressor(n_estimators=100, random_state=42)),
        ('en', ElasticNet(random_state=42))
    ]))
])

# Entrenar ensembles
ensemble_ron.fit(X_train, y_ron_train)
ensemble_mon.fit(X_train, y_mon_train)

# Evaluar ensembles
ron_ensemble_pred = ensemble_ron.predict(X_test)
mon_ensemble_pred = ensemble_mon.predict(X_test)

ron_ensemble_r2 = r2_score(y_ron_test, ron_ensemble_pred)
ron_ensemble_rmse = np.sqrt(mean_squared_error(y_ron_test, ron_ensemble_pred))
mon_ensemble_r2 = r2_score(y_mon_test, mon_ensemble_pred)
mon_ensemble_rmse = np.sqrt(mean_squared_error(y_mon_test, mon_ensemble_pred))

print(f"Ensemble:")
print(f"  RON - R²: {ron_ensemble_r2:.4f}, RMSE: {ron_ensemble_rmse:.4f}")
print(f"  MON - R²: {mon_ensemble_r2:.4f}, RMSE: {mon_ensemble_rmse:.4f}")

# =============================================================================
# VALIDACIÓN CRUZADA ROBUSTA
# =============================================================================
print("\n=== VALIDACIÓN CRUZADA ===")

# Usar el mejor modelo individual para validación cruzada
best_pipeline_ron = pipelines_ron[best_ron_model]
best_pipeline_mon = pipelines_mon[best_mon_model]

cv_scores_ron = cross_val_score(best_pipeline_ron, X, y_ron, cv=5, scoring='r2')
cv_scores_mon = cross_val_score(best_pipeline_mon, X, y_mon, cv=5, scoring='r2')

print(f"Mejor modelo RON ({best_ron_model}): R² CV = {cv_scores_ron.mean():.4f} (+/- {cv_scores_ron.std() * 2:.4f})")
print(f"Mejor modelo MON ({best_mon_model}): R² CV = {cv_scores_mon.mean():.4f} (+/- {cv_scores_mon.std() * 2:.4f})")

# =============================================================================
# MODELO NO SUPERVISADO MEJORADO
# =============================================================================
print("\n=== MODELO NO SUPERVISADO MEJORADO ===")

# Preprocesamiento para PCA
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# PCA óptimo
pca = PCA()
X_pca = pca.fit_transform(X_scaled)

# Determinar número óptimo de componentes
explained_variance = np.cumsum(pca.explained_variance_ratio_)
optimal_components = np.argmax(explained_variance >= 0.95) + 1

print(f"Componentes para 95% de varianza: {optimal_components}")
print(f"Varianza explicada: {explained_variance[optimal_components-1]:.3f}")

# Recalcular PCA con componentes óptimos
pca_optimal = PCA(n_components=optimal_components)
X_pca_optimal = pca_optimal.fit_transform(X_scaled)

# Determinar número óptimo de clusters
print("\n--- Determinando Número Óptimo de Clusters ---")
silhouette_scores = []
for k in range(2, 8):
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    cluster_labels = kmeans.fit_predict(X_pca_optimal)
    silhouette_avg = silhouette_score(X_pca_optimal, cluster_labels)
    silhouette_scores.append(silhouette_avg)
    print(f"Clusters: {k}, Silhouette Score: {silhouette_avg:.3f}")

optimal_clusters = np.argmax(silhouette_scores) + 2
print(f"Número óptimo de clusters: {optimal_clusters}")

# Clustering con número óptimo
kmeans_optimal = KMeans(n_clusters=optimal_clusters, random_state=42, n_init=10)
cluster_labels = kmeans_optimal.fit_predict(X_pca_optimal)

# =============================================================================
# ANÁLISIS DE RESULTADOS
# =============================================================================
print("\n" + "="*50)
print("RESULTADOS MEJORADOS")
print("="*50)

print("\n=== MODELO SUPERVISADO MEJORADO ===")
print(f"RON - Mejor modelo: {best_ron_model}")
print(f"RON - R² Test: {best_ron_score:.4f}")
print(f"RON - R² CV: {cv_scores_ron.mean():.4f}")
print(f"RON - RMSE: {np.sqrt(mean_squared_error(y_ron_test, pipelines_ron[best_ron_model].predict(X_test))):.4f}")

print(f"\nMON - Mejor modelo: {best_mon_model}")
print(f"MON - R² Test: {best_mon_score:.4f}")
print(f"MON - R² CV: {cv_scores_mon.mean():.4f}")
print(f"MON - RMSE: {np.sqrt(mean_squared_error(y_mon_test, pipelines_mon[best_mon_model].predict(X_test))):.4f}")

print(f"\nEnsemble:")
print(f"RON - R²: {ron_ensemble_r2:.4f}, RMSE: {ron_ensemble_rmse:.4f}")
print(f"MON - R²: {mon_ensemble_r2:.4f}, RMSE: {mon_ensemble_rmse:.4f}")

print("\n=== MODELO NO SUPERVISADO MEJORADO ===")
print(f"Components PCA: {optimal_components}")
print(f"Varianza explicada: {explained_variance[optimal_components-1]:.3f}")
print(f"Clusters óptimos: {optimal_clusters}")
print(f"Mejor Silhouette Score: {max(silhouette_scores):.3f}")

# Análisis de clusters
df_clusters = df.copy()
df_clusters['Cluster'] = cluster_labels

print(f"\nRelación Clusters - RON/MON:")
cluster_stats = df_clusters.groupby('Cluster').agg({
    'RON': ['mean', 'std'],
    'MON': ['mean', 'std']
}).round(6)

print(cluster_stats)

# =============================================================================
# VISUALIZACIONES MEJORADAS
# =============================================================================
print("\n=== GENERANDO VISUALIZACIONES ===")

plt.figure(figsize=(15, 10))

# 1. Varianza explicada por PCA
plt.subplot(2, 3, 1)
plt.plot(range(1, len(explained_variance) + 1), explained_variance, 'bo-')
plt.axhline(y=0.95, color='r', linestyle='--', label='95% varianza')
plt.axvline(x=optimal_components, color='g', linestyle='--', label=f'Optimal: {optimal_components}')
plt.xlabel('Número de Componentes')
plt.ylabel('Varianza Explicada Acumulada')
plt.title('Varianza Explicada por PCA')
plt.legend()
plt.grid(True)

# 2. Silhouette scores
plt.subplot(2, 3, 2)
plt.plot(range(2, 8), silhouette_scores, 'ro-')
plt.axvline(x=optimal_clusters, color='g', linestyle='--', label=f'Óptimo: {optimal_clusters}')
plt.xlabel('Número de Clusters')
plt.ylabel('Silhouette Score')
plt.title('Análisis de Silhouette')
plt.legend()
plt.grid(True)

# 3. Clusters en espacio PCA
plt.subplot(2, 3, 3)
scatter = plt.scatter(X_pca_optimal[:, 0], X_pca_optimal[:, 1], c=cluster_labels, cmap='viridis')
plt.xlabel(f'PC1 ({pca_optimal.explained_variance_ratio_[0]:.2%})')
plt.ylabel(f'PC2 ({pca_optimal.explained_variance_ratio_[1]:.2%})')
plt.title('Clusters en Espacio PCA')
plt.colorbar(scatter)

# 4. Comparación de modelos RON
plt.subplot(2, 3, 4)
models_names = ['RF', 'GB', 'EN', 'Ensemble']
ron_scores = [pipelines_ron['RandomForest'].score(X_test, y_ron_test),
             pipelines_ron['GradientBoosting'].score(X_test, y_ron_test),
             pipelines_ron['ElasticNet'].score(X_test, y_ron_test),
             ron_ensemble_r2]
plt.bar(models_names, ron_scores, color=['blue', 'orange', 'green', 'red'])
plt.ylim(0.9, 1.0)
plt.ylabel('R² Score')
plt.title('Comparación Modelos - RON')

# 5. Comparación de modelos MON
plt.subplot(2, 3, 5)
mon_scores = [pipelines_mon['RandomForest'].score(X_test, y_mon_test),
             pipelines_mon['GradientBoosting'].score(X_test, y_mon_test),
             pipelines_mon['ElasticNet'].score(X_test, y_mon_test),
             mon_ensemble_r2]
plt.bar(models_names, mon_scores, color=['blue', 'orange', 'green', 'red'])
plt.ylim(0.9, 1.0)
plt.ylabel('R² Score')
plt.title('Comparación Modelos - MON')

# 6. Distribución de RON/MON por cluster
plt.subplot(2, 3, 6)
for cluster in range(optimal_clusters):
    cluster_data = df_clusters[df_clusters['Cluster'] == cluster]
    plt.scatter(cluster_data['RON'], cluster_data['MON'],
               label=f'Cluster {cluster}', alpha=0.7)
plt.xlabel('RON')
plt.ylabel('MON')
plt.title('Distribución RON vs MON por Cluster')
plt.legend()

plt.tight_layout()
plt.show()

# =============================================================================
# RESUMEN FINAL
# =============================================================================
print("\n" + "="*50)
print("RESUMEN DE MEJORAS IMPLEMENTADAS")
print("="*50)
print("✅ Validación cruzada robusta")
print("✅ Ensemble de modelos")
print("✅ Selección automática de características")
print("✅ Determinación óptima de componentes PCA")
print("✅ Análisis de silhouette para clusters")
print("✅ Pipelines reproducibles")
print("✅ Visualizaciones mejoradas")
print("✅ Evitación de sobreajuste")

print(f"\nMejora esperada en R²: {min(ron_ensemble_r2, mon_ensemble_r2):.3f} - {max(ron_ensemble_r2, mon_ensemble_r2):.3f}")
```
